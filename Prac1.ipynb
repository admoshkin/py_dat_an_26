```python
class Chocolate:

  default_choc = "Обычный"

  def __init__(self, cost, supp: str = default_choc):

    self.cost = cost
    self.supp = supp

    if cost <= 0:
      raise ValueError('Цена должна быть положительной')

  def show_my_chocolate(self):
    if self.supp == self.default_choc:
      print('Обычный шоколадный батончик')

    else:
      print(f'Шоколадный батончик, добавка: {self.supp}')



  def __str__(self) -> str:
    return f'Шоколадный батончик, {self.cost} рублей'
```

```python
babaevsky = Chocolate(70)
print(babaevsky)
babaevsky.show_my_chocolate()
```

```python
baby_fox = Chocolate(45,'молочная начинка')
print(baby_fox)
baby_fox.show_my_chocolate()
```
